```python
import math
```

```python
def f(x):
    return math.sqrt((abs(x - 2)) ** 5)
```

```python
a, b = 0, 5
epsilon = 1e-10
delta = epsilon*10
```

```python
def dichotomy(f, a, b):
    iterations = 0
    calculations = 0
    while (b - a) > delta:

        x1 = (a + b - epsilon) / 2
        x2 = (a + b + epsilon) / 2

        f1 = f(x1)
        f2 = f(x2)

        calculations += 2
        iterations += 1

        if f1 < f2:
            b = x2
        else:
            a = x1
    return (a + b) / 2, b - a, iterations, calculations
```

```python
def golden_section(f, a, b):

    iterations = 0
    calculations = 0
    multiplier = (math.sqrt(5) - 1) / 2  # 0.618

    x1 = b - (b - a) * multiplier
    x2 = a + (b - a) * multiplier

    f1 = f(x1)
    f2 = f(x2)

    calculations += 2

    while (b - a) > delta:

        iterations += 1
        calculations += 1

        if f1 < f2:
            b = x2
            x2 = x1
            f2 = f1
            x1 = b - multiplier * (b - a)
            f1 = f(x1)
        else:
            a = x1
            x1 = x2
            f1 = f2
            x2 = a + multiplier * (b - a)
            f2 = f(x2)

    return (a + b) / 2, b - a, iterations, calculations
```

```python
def fibonachi(f, a, b, n):
    fib = [1, 1]
    for _ in range(n):
        fib.append(fib[-1] + fib[-2])

    iterations = 0
    calculations = 0

    x1 = a + (fib[n - 2] / fib[n]) * (b - a)
    x2 = a + (fib[n - 1] / fib[n]) * (b - a)

    f1 = f(x1)
    f2 = f(x2)

    calculations += 2

    for k in range(1, n - 2):
        calculations += 1
        iterations += 1
        if f1 < f2:
            b = x2
            x2 = x1
            f2 = f1
            x1 = a + (fib[n - k - 1] / fib[n - k + 1]) * (b - a)
            f1 = f(x1)
        else:
            a = x1
            x1 = x2
            f1 = f2
            x2 = a + (fib[n - k] / fib[n - k + 1]) * (b - a)
            f2 = f(x2)

    if f1 < f2:
        b = x2
        x2 = x1
        f2 = f1
        x_new = (x1 + b - epsilon) / 2
        f_new = f(x_new)
        calculations += 1
        
        if f1 < f_new:
            b = x_new
        else:
            a = x1
    else:
        a = x1
        x1 = x2
        f1 = f2
        x_new = (a + x2 + epsilon) / 2 
        f_new = f(x_new)
        calculations += 1
        
        if f_new < f2:
            a = x_new
        else:
            b = x2

    return (a + b) / 2, b - a, iterations, calculations
```

```python
(
    dichotomy_result,
    dichotomy_interval_length,
    dichotomy_iterations,
    dichotomy_calculations,
) = dichotomy(f, a, b)
```

```python
(
    golden_section_result,
    golden_section_interval_length,
    golden_section_iterations,
    golden_section_calculations,
) = golden_section(f, a, b)
```

```python
(
    fibonachi_result,
    fibonachi_interval_length,
    fibonachi_iterations,
    fibonachi_calculations,
) = fibonachi(f, a, b, golden_section_calculations)
```
