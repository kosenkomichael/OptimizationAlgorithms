```python
import math
```

```python
def f(x):
    return math.sqrt((abs(x - 2)) ** 5)
```

```python
a, b = 0, 5
delta = 1e-9
epsilon = 1e-10
```

```python
def dichotomy(f,a,b):
    iterations = 0
    calculations = 0
    while (b - a) > delta:
        
        x1 = (a + b - epsilon) / 2
        x2 = (a + b + epsilon) / 2

        f1 = f(x1)
        f2 = f(x2)

        calculations += 2
        iterations += 1

        if f1 < f2:
            b = x2
        else:
            a = x1
    return ( a + b ) / 2, b - a, iterations, calculations
```

```python
def golden_section(f,a,b):

    iterations = 0
    calculations = 0
    multiplier = 0.618

    x1 = b - (b - a) * multiplier
    x2 = a + (b - a) * multiplier

    f1 = f(x1)
    f2 = f(x2)

    calculations += 2
    iterations += 1

    while (b - a) > delta:

        iterations += 1
        calculations += 1
        
        if f1 < f2:
            b = x2
            x2 = x1
            f2 = f1
            x1 = b - multiplier * (b - a)
            f1 = f(x1)
        else:
            a = x1
            x1 = x2
            f1 = f2
            x2 = a + multiplier * (b - a)
            f2 = f(x2)

    return ( a + b ) / 2, b - a, iterations, calculations
```

```python
def fibonachi(f,a,b,n):
    fib = [1,1]
    for _ in range(n):
        fib.append(fib[-1] + fib[-2])

    iterations = 0
    calculations = 0

    x1 = a + (fib[n - 2] / fib[n]) * (b - a)
    x2 = a + (fib[n - 1] / fib[n]) * (b - a)

    f1 = f(x1)
    f2 = f(x2)

    calculations += 2
    iterations += 1

    for k in range(2,n):
        calculations += 1
        iterations += 1
        if f1 < f2:
            b = x2
            x2 = x1
            f2 = f1
            x1 = a + (fib[n - k - 1] / fib[n - k + 1]) * (b - a)
            f1 = f(x1)
        else:
            a = x1
            x1 = x2
            f1 = f2
            x2 = a + (fib[n - k] / fib[n - k + 1]) * (b - a)
            f2 = f(x2)

    x1_final = (a + b) / 2 - epsilon
    x2_final = (a + b) / 2 + epsilon

    f1_final = f(x1_final)
    f2_final = f(x2_final)

    iterations += 1
    calculations += 2

    if f1_final < f2_final:
        x_min = x1_final
        b = x2_final
    else:
        x_min = x2_final
        a = x1_final

    return x_min, b - a, iterations, calculations
```

```python
dichotomy_result,dichotomy_interval_length, dichotomy_iterations, dichotomy_calculations = dichotomy(f,a,b)
```

```python
golden_section_result,golden_section_interval_length, golden_section_iterations, golden_section_calculations = golden_section(f,a,b)
```

```python
fibonachi_result,fibonachi_interval_length, fibonachi_iterations, fibonachi_calculations = fibonachi(f,a,b, golden_section_calculations - 2)
```

```python
print('-' * 70)
print("Точность (аналитический ответ = 2)")
print("dichotomy     ", dichotomy_result, abs(dichotomy_result - 2) / 2)
print("golden_section", golden_section_result, abs(golden_section_result - 2) / 2)
print("fibonachi     ", fibonachi_result, abs(fibonachi_result - 2) / 2)
print('-' * 70)
print("итоговый интервал")
print("dichotomy     ", dichotomy_interval_length)
print("golden_section", golden_section_interval_length)
print("fibonachi     ", fibonachi_interval_length)
print('-' * 70)
print("число итераций")
print("dichotomy     ", dichotomy_iterations)
print("golden_section", golden_section_iterations)
print("fibonachi     ", fibonachi_iterations)
print('-' * 70)
print("число вычислений f(x)")
print("dichotomy     ", dichotomy_calculations)
print("golden_section", golden_section_calculations)
print("fibonachi     ", fibonachi_calculations)
print('-' * 70)
print("Вычислений у золотого сечения меньше, чем у дихотомии", golden_section_calculations < dichotomy_calculations)
print("Точность фибоначчи лучше, чем у золотого сечения     ", abs(fibonachi_result - 2) / 2 < abs(golden_section_result - 2) / 2)
```
